## Data Strategy Notes
- 1- Files smaller than 100 MB will be loaded normally using Pandas.
- 2- Medium-sized files (>100 MB) will be assessed for memory optimization when necessary.
- 3- Large files (>500 MB) require memory optimization techniques such as:
    - category dtype
    - bool dtype
    - usecols
    - datetime parsing
- 4- Extremely large files (>1 GB) will be processed using:
      - Chunksize
      - Parquet Files
      - Aggregations when possible
- 5- Never load more data than needed.
- 6- Required columns will be loaded based on the business question whenever possible.
