### **category distribution**

```python
products_by_category = (
    products
        .groupby("category")
        .agg(
            products_count = ("product_id", "count")
        )
        .sort_values("products_count", ascending = False)
)
products_by_category["perccentage"] = (products_by_category["products_count"] / products_by_category["products_count"].sum()).round(4)
products_by_category.to_csv("../../output/products_by_category.csv")
products_by_category
```
